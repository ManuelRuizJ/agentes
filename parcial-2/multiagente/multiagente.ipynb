````python
# Función para cargar base de datos
def cargar_ventas():
    df = pd.read_csv("/content/drive/MyDrive/Anahuac/IA/databases/ventas.csv")
    return df
````

````python
# Función para calcular el total de ventas
def total_ventas(df=None):
    if df is None:
        df = cargar_ventas()
    total = df["ventas"].sum()
    return total
````

````python
# Función para calcular los productos mas vendidos
def top_productos(df, n=5):
    if df is None:
        df = cargar_ventas()
    resultado = (
        df.groupby("producto")
        ["ventas"]
        .sum()
        .sort_values(ascending=False)
    )
    return resultado.head(n)
````

````python
# Función para identificar las sucursales con mayores ventas
def top_sucursales(df, n=5):
    if df is None:
        df = cargar_ventas()
    resultado = (
        df.groupby("sucursal")
        ["ventas"]
        .sum()
        .sort_values(ascending=False)
    )
    return resultado.head(n)
````

### Nuevas Funciones Propuestas

````python
def top_clientes_por_gasto(df_clientes_arg=None, n=5):
    """
    Identifica los clientes que más han gastado.
    """
    if df_clientes_arg is None:
        # Asume que df_clientes ya está cargado globalmente o debe ser cargado
        # Para este ejemplo, usaremos la variable global df_clientes
        global df_clientes
        if 'df_clientes' not in globals():
            raise ValueError("df_clientes no está cargado. Asegúrate de cargarlo antes o pásalo como argumento.")
        df_clientes_arg = df_clientes

    resultado = df_clientes_arg.sort_values(by='total_compras', ascending=False)
    return resultado.head(n)

````

Esta función, `top_clientes_por_gasto`, toma el DataFrame `df_clientes` y devuelve una lista de los `n` clientes que han realizado las mayores compras en términos de `total_compras`.

````python
def productos_bajo_reorden(df_inventarios_arg=None):
    """
    Lista los productos cuyo stock actual está por debajo del punto de reorden.
    """
    if df_inventarios_arg is None:
        # Asume que df_inventarios ya está cargado globalmente o debe ser cargado
        global df_inventarios
        if 'df_inventarios' not in globals():
            raise ValueError("df_inventarios no está cargado. Asegúrate de cargarlo antes o pásalo como argumento.")
        df_inventarios_arg = df_inventarios

    productos_en_riesgo = df_inventarios_arg[df_inventarios_arg['existencia'] < df_inventarios_arg['punto_reorden']]
    return productos_en_riesgo[['producto', 'existencia', 'punto_reorden']]
````

La función `productos_bajo_reorden` examina el DataFrame `df_inventarios` y filtra los productos cuya cantidad en `existencia` es menor que su `punto_reorden`, indicando posibles necesidades de reposición.

````python
def ventas_mensuales(df_ventas_arg=None):
    """
    Calcula el total de ventas por mes.
    """
    if df_ventas_arg is None:
        global df_ventas
        if 'df_ventas' not in globals():
            raise ValueError("df_ventas no está cargado. Asegúrate de cargarlo antes o pásalo como argumento.")
        df_ventas_arg = df_ventas

    # Asegurarse de que la columna 'fecha' sea de tipo datetime
    df_ventas_arg['fecha'] = pd.to_datetime(df_ventas_arg['fecha'])
    # Extraer el mes y agrupar
    ventas_por_mes = df_ventas_arg.groupby(df_ventas_arg['fecha'].dt.to_period('M'))['ventas'].sum()
    return ventas_por_mes
````

La función `ventas_mensuales` procesa el DataFrame `df_ventas` para calcular la suma total de ventas para cada mes presente en los datos. Esto ayuda a identificar tendencias estacionales o mensuales en las ventas.

````python
# -------- Nuevos Tools ---------

@tool
def obtener_top_clientes_por_gasto_tool():
    """
    Obtiene los clientes que más han gastado.
    """
    return top_clientes_por_gasto().to_string()

@tool
def obtener_productos_bajo_reorden_tool():
    """
    Lista los productos cuyo stock actual está por debajo del punto de reorden.
    """
    return productos_bajo_reorden().to_string()

@tool
def obtener_ventas_mensuales_tool():
    """
    Calcula el total de ventas por mes.
    """
    return ventas_mensuales().to_string()

print("Nuevos tools para clientes, inventario y ventas mensuales definidos.")
````

````python
# --------- Definición de skills ---------
@tool
def obtener_top_productos():
    """
    Obtiene los productos más vendidos
    """
    return top_productos(None).to_string()

@tool
def obtener_top_sucursales():
    """
    Obtiene las sucursales con mayores ventas
    """
    return top_sucursales(None).to_string()
````

````python
# --------- Definición de agentes ---------
analista = Agent(
    role="Analista de Ventas",
    goal="Analizar información comercial y generar un resumen ejecutivo",
    backstory="Especialista en inteligencia de negocios con experiencia en análisis de ventas",
    tools=[obtener_top_productos, obtener_top_sucursales],
    llm=llm_ollama,
    verbose=True  # Para ver qué está haciendo
)
````

````python
# --------- Definición de agentes adicionales ---------

analista_clientes = Agent(
    role="Analista de Clientes",
    goal="Identificar y analizar los clientes más valiosos y sus patrones de compra.",
    backstory="Experto en comportamiento del consumidor y análisis de datos de clientes, buscando maximizar el valor del cliente.",
    tools=[obtener_top_clientes_por_gasto_tool],
    llm=llm_ollama,
    verbose=True
)

analista_inventarios = Agent(
    role="Analista de Inventario",
    goal="Gestionar y optimizar los niveles de inventario, identificando riesgos de stock bajo.",
    backstory="Especialista en logística y cadena de suministro, enfocado en mantener un inventario eficiente.",
    tools=[obtener_productos_bajo_reorden_tool],
    llm=llm_ollama,
    verbose=True
)

analista_ventas_mensuales = Agent(
    role="Analista de Tendencias de Ventas",
    goal="Analizar las ventas mensuales para identificar patrones y tendencias estacionales o de crecimiento.",
    backstory="Especialista en el análisis de series de tiempo de ventas, detectando oportunidades y riesgos en el rendimiento mensual.",
    tools=[obtener_ventas_mensuales_tool],
    llm=llm_ollama,
    verbose=True
)

administrador_orquestador = Agent(
    role="Administrador Orquestador de Informes",
    goal="Coordinar a los agentes especializados para generar informes comerciales completos y precisos basados en la solicitud del usuario.",
    backstory="Soy el cerebro detrás de la generación de informes, seleccionando y delegando tareas a los analistas adecuados para responder a las preguntas más críticas del negocio.",
    tools=[
        obtener_top_productos,
        obtener_top_sucursales,
        obtener_top_clientes_por_gasto_tool,
        obtener_productos_bajo_reorden_tool,
        obtener_ventas_mensuales_tool
    ], # El orquestador tiene acceso a todas las herramientas de datos
    llm=llm_ollama,
    verbose=True
)

redactor_informe_final = Agent(
    role="Redactor de Informes Ejecutivos",
    goal="Compilar, sintetizar y presentar de manera clara y concisa los hallazgos de los agentes en un informe ejecutivo final.",
    backstory="Convierto datos complejos y análisis detallados en informes ejecutivos comprensibles y accionables para la toma de decisiones.",
    llm=llm_ollama,
    verbose=True
)

print("Agentes adicionales configurados: analista_clientes, analista_inventarios, analista_ventas_mensuales, administrador_orquestador y redactor_informe_final.")
````

````python
# --------- Creación de tareas ---------
tarea = Task(
    description="""
    Analiza los productos más vendidos usando la herramienta obtener_top_productos.
    Genera un resumen ejecutivo con los hallazgos principales.
    """,
    expected_output="""
    Un resumen ejecutivo en español que incluya:
    - Lista de los productos más vendidos
    - Análisis breve de los resultados
    """,
    agent=analista
)
````

````python
# --------- Creación de tareas adicionales y orquestación ---------

tarea_orquestador = Task(
    description=(
        "Basado en la solicitud del usuario de generar un informe, "
        "utiliza las herramientas disponibles para recopilar toda la información relevante. "
        "La solicitud del usuario es: {report_request}. "
        "Asegúrate de ejecutar las herramientas necesarias para obtener los datos crudos o resúmenes que el redactor necesite."
    ),
    expected_output=(
        "Un resumen detallado y estructurado de los datos solicitados por el usuario, "
        "listo para ser procesado por el redactor del informe. "
        "Si se solicitan varios tipos de informes, incluye la información de todos ellos de manera clara. "
        "Por ejemplo, si se pide 'informe de ventas y clientes', incluye top productos, top sucursales y top clientes."
    ),
    agent=administrador_orquestador
)

tarea_redactor_informe = Task(
    description=(
        "Toma la información generada por el Administrador Orquestador y redacta un informe ejecutivo final. "
        "El informe debe ser claro, conciso, estar bien estructurado y en español, cubriendo todos los puntos de la solicitud original. "
        "Incluye encabezados claros para cada sección y un breve análisis de los hallazgos. "
        "El informe debe ser adecuado para ser presentado a la dirección."
    ),
    expected_output=(
        "Un informe ejecutivo final en formato Markdown, con un título, "
        "secciones claras para cada tipo de análisis (e.g., 'Análisis de Ventas', 'Análisis de Clientes', 'Análisis de Inventario'), "
        "y un resumen de los hallazgos clave. "
        "El informe debe ser legible y profesional, listo para su almacenamiento y presentación."
    ),
    agent=redactor_informe_final,
    context=[tarea_orquestador] # El redactor usará el output del orquestador como contexto
)

print("Tareas de orquestación y redacción de informe final definidas.")
````

````python
# --------- Creación de Crew ---------
crew = Crew(
    name="Analista de Ventas",
    description="Analiza información comercial",
    agents=[analista],
    tasks=[tarea],
    verbose=True
)
````

````python
# --------- Creación de la Crew con Orquestador y Redactor ---------
# La crew ahora incluye al orquestador para decidir qué herramientas usar
# y al redactor para formatear la salida final.

crew = Crew(
    name="Generador de Informes Comerciales",
    description="Genera informes comerciales ejecutivos personalizados a petición del usuario, utilizando agentes especializados.",
    agents=[administrador_orquestador, redactor_informe_final], # Solo el orquestador y el redactor en la crew principal
    tasks=[tarea_orquestador, tarea_redactor_informe], # Las tareas se ejecutan secuencialmente
    verbose=True
)

print("Crew 'Generador de Informes Comerciales' configurada.")
````

````python
# --------- Ejecución de la Crew ---------
import asyncio

print("📊 Ejecutando análisis...")

async def run_crew():
    try:
        resultado = await crew.kickoff_async()
        print("\nResultado de la Crew:")
        print(resultado)
    except Exception as e:
        print(f"Error: {e}")
        print("\nDiagnóstico:")
        print(f"- URL de Ollama: {OLLAMA_BASE_URL}")
        print(f"- ¿Está corriendo Ollama? Verifica en la terminal: ollama serve")

await run_crew()
````

## Menú de Informes y Ejecución

Selecciona el tipo de informe que deseas generar. El Administrador Orquestador decidirá qué agentes y herramientas utilizar para recopilar la información y el Redactor de Informes Final la compilará en un documento ejecutivo.

````python
# --------- Menú de selección de informe y ejecución ---------

async def run_report_generator():
    print("\n--- Menú de Generación de Informes ---")
    print("1. Informe de Ventas (Productos y Sucursales)")
    print("2. Informe de Clientes (Top Clientes por Gasto)")
    print(
        "3. Informe de Inventario (Productos bajo Reorden)"
    )
    print("4. Informe de Ventas Mensuales")
    print("5. Informe Completo (Ventas, Clientes, Inventario, Ventas Mensuales)")
    print("6. Salir")

    while True:
        choice = input("Ingrese su elección (1-6): ")
        report_request = ""

        if choice == '1':
            report_request = "Genera un informe detallado de los productos más vendidos y las sucursales con mayores ventas."
        elif choice == '2':
            report_request = "Genera un informe sobre los clientes que más han gastado."
        elif choice == '3':
            report_request = "Genera un informe de los productos cuyo stock está por debajo del punto de reorden."
        elif choice == '4':
            report_request = "Genera un informe de las ventas totales por mes para identificar tendencias."
        elif choice == '5':
            report_request = "Genera un informe ejecutivo completo que incluya los productos más vendidos, las sucursales con mayores ventas, los clientes que más han gastado, los productos bajo reorden y el total de ventas mensuales."
        elif choice == '6':
            print("Saliendo del generador de informes.")
            return
        else:
            print("Elección no válida. Por favor, intente de nuevo.")
            continue

        print(f"\n📊 Generando informe para: {report_request}")
        try:
            # Ejecutar la crew, pasando la solicitud del usuario como entrada para la primera tarea
            result = await crew.kickoff_async(inputs={'report_request': report_request})

            print("\n--- Resultado Final del Informe ---")
            print(result)

            # Guardar el informe en Google Drive
            report_filename = f"informe_{choice}_{pd.Timestamp.now().strftime('%Y%m%d_%H%M%S')}.md"
            report_filepath = os.path.join(report_dir, report_filename)
            with open(report_filepath, "w", encoding="utf-8") as f:
                f.write(result)
            print(f"Informe guardado en: {report_filepath}")

        except Exception as e:
            print(f"Error durante la ejecución de la Crew: {e}")
            print("\nDiagnóstico:")
            print(f"- URL de Ollama: {OLLAMA_BASE_URL}")
            print(f"- ¿Está corriendo Ollama? Verifica en la terminal: ollama serve")
        print("\n-----------------------------------\n")

# Ejecutar el generador de informes
await run_report_generator()
````
